```python
# Guessing game
import random
num = random.randint(1, 100)
guess_num = int(input('Guess the number: '))
count = 1

while guess_num != num:
    if guess_num < num:
        print('Guess a larger number')
        count += 1
        guess_num = int(input('Guess the number:'))

    else:
        print('Guess a smaller number')
        count += 1
        guess_num = int(input('Guess the number: '))

else:
    print('You guess the number in', count, 'times')
    print(num)
```

#### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
current_population = 10000
print("current_population: ", current_population)
current_population = current_population / 1.1

for i in range(1, 10, 1):
    print(f'{i} year ago: {round(current_population)}')
    current_population = current_population / 1.1
```

```python
# print prime number
lower = int(input('Enter lower range: '))
upper = int(input('Enter upper range: '))

for i in range(lower, upper+1):
    for j in range(2, i):
        if i % j == 0:
            break
    else:
        print(i)
```

```python
# Write a program which can remove a particular character from a string.
s = input('Enter the string: ')
remove_char = input('Enter the char you want to remove: ')
new_str = ''

for i in s:
    if i != remove_char:
        new_str += i
print(new_str)
```

```python
# Write a program to count the number of words in a string without split()
s = input('Enter String: ')
# ls = s.split(' ')
# print('Total numbers of words in a string', len(ls))

l = []
temp = ''
for i in s:
  if i != ' ':
      temp += i
  else:
      l.append(temp)
      temp = ''
l.append(temp)
print(l)

cnt = 0
for j in l:
  cnt += 1
print('Total numbers of words in a string', cnt)
```

```python
# Write a python program to convert a string to title case without using the title()
s = input('Enter String: ')
# print(s.title())
lst = []
l = s.split(' ')
for i in l:
  lst.append(i[0].upper() + i[1:].lower())

print(' '.join(lst))
```

```python
# Nested if with List Comprehension
basket = ['apple','guava','cherry','banana']
my_fruits = ['apple','kiwi','grapes','banana']

# add new list from my_fruits and items if the fruit exists in basket and also starts with 'a'

print([fruit for fruit in my_fruits if fruit in basket if fruit.startswith('a')])

print([fruit for fruit in my_fruits if fruit in basket and fruit.startswith('a')])
```

```python
# Print a (3,3) matrix using list comprehension -> Nested List comprehension
[[i*j for i in range(1,4)] for j in range(1,4)]
```

```python
# cartesian products -> List comprehension on 2 lists together
L1 = [1,2,3,4]
L2 = [5,6,7,8]

[i*j for i in L1 for j in L2]
```
