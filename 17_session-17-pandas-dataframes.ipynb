```python
# write a function that can return the track record of 2 teams against each other
T1 = (ipl['Team1'] == 'Mumbai Indians') & (ipl['Team2'] == 'Chennai Super Kings')
ipl[T1]
T2 = (ipl['Team1'] == 'Chennai Super Kings') & (ipl['Team2'] == 'Mumbai Indians')
ipl[T2]
print("Total Match:", ipl[T1].shape[0] + ipl[T2].shape[0])
new_df1 = ipl[T1]
new_df2 = ipl[T2]
new_df1['WinningTeam'].value_counts() + new_df2['WinningTeam'].value_counts()
```

```python
def TrackRecord(T1, T2):
    team1 = (ipl['Team1'] == T1) & (ipl['Team2'] == T2)
    team2 = (ipl['Team1'] == T2) & (ipl['Team2'] == T1)

    new_df1 = ipl[team1]
    new_df2 = ipl[team2]

    print("Total Match:", new_df1.shape[0] + new_df2.shape[0])
    print(new_df1['WinningTeam'].value_counts() + new_df2['WinningTeam'].value_counts())
```

```python
TrackRecord('Mumbai Indians', 'Chennai Super Kings')
```

```python
TrackRecord('Chennai Super Kings', 'Mumbai Indians')
```

```python
TrackRecord('Chennai Super Kings', 'Kolkata Knight Riders')
```
